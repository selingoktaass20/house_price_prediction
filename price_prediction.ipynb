```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("processed_turkish_house_sales.csv")
df
```

```python
df_istanbul = df[df["il"] == "Istanbul"]
```

```python
df_istanbul
```

```python
df_istanbul["satici_tip"].value_counts()
```

```python
df_istanbul["satici_tip"] = df_istanbul["satici_tip"].map({
    "Sahibinden": 0,
    "Emlak-ofisinden": 1
})
df_istanbul
```

```python
df_istanbul["Oda_Sayisi"].value_counts()
```

```python
counts = df_istanbul["Oda_Sayisi"].value_counts()
valid_rooms = counts[counts >= 5].index
valid_rooms
```

```python
df_istanbul = df_istanbul[df_istanbul["Oda_Sayisi"].isin(valid_rooms)]
df_istanbul
```

```python
df_istanbul = df_istanbul.drop("il", axis = 1)
df_istanbul
```

```python
df_istanbul["Ilce"].value_counts()
```

```python
df_istanbul["Mahalle"].value_counts()
```

```python
mahalle = df_istanbul["Mahalle"].value_counts()
nadir_mahalle = mahalle[mahalle<4]
df_istanbul["Mahalle"] = df_istanbul["Mahalle"].replace(nadir_mahalle.index, "diger")
df_istanbul["Mahalle"].value_counts()
```

```python
df_istanbul["Tarih"].value_counts()
```

```python
df_istanbul = df_istanbul.drop("Tarih", axis = 1)
df_istanbul
```

```python
df_istanbul = pd.get_dummies(
    df_istanbul,
    columns=["Oda_Sayisi", "Ilce", "Mahalle"],
    drop_first=True,
    dtype=int
)
df_istanbul
```

```python
df_istanbul["Metrekare"].skew()
```

```python
df_istanbul["Metrekare"] = np.log1p(df_istanbul["Metrekare"])
```

```python
df_istanbul["fiyat"].skew()
```

```python
df_istanbul["fiyat"] = np.log1p(df_istanbul["fiyat"])
```

```python
X = df_istanbul.drop("fiyat", axis = 1)
Y = df_istanbul["fiyat"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
model = RandomForestRegressor(n_estimators=200, max_depth=8)
model.fit(x_train, y_train)
tahmin = model.predict(x_test)
mse = mean_squared_error(y_test, tahmin)
r2 = r2_score(y_test, tahmin)
mse, r2
```

```python
from xgboost import XGBRegressor
```

```python
xgb = XGBRegressor(n_estimators=200, max_depth=8)
xgb.fit(x_train, y_train)
tahmin = xgb.predict(x_test)
mse = mean_squared_error(y_test, tahmin)
r2 = r2_score(y_test, tahmin)
mse, r2
```
